```python
def generate_booking_id(issued_ids: set) -> str:
    
    counter = 1001
    while True:
        candidate_id = f"B{counter}"
        if candidate_id not in issued_ids:
            issued_ids.add(candidate_id)
            return candidate_id
        counter += 1

# Quick test of ID generator
test_ids = set()
sample_id_1 = generate_booking_id(test_ids)
sample_id_2 = generate_booking_id(test_ids)
print(f"Generated IDs: {sample_id_1}, {sample_id_2} (Unique set size: {len(test_ids)})")
```

 # Search for a Route  
### FEATURE 04

```python
def search_route(route_list, search_term):
    found = False

    print(f"\nSearch results for: '{search_term}'")
    print("*" * 65)

    for r in route_list:
        if (search_term.lower() == r["route_id"].lower()
            or search_term.lower() in r["departure_city"].lower()
            or search_term.lower() in r["destination_city"].lower()):

            print(f"\nRoute ID: {r['route_id']}")
            print(f"From: {r['departure_city']}")
            print(f"To: {r['destination_city']}")
            print(f"Departure: {r['departure_time']}")
            print(f"Ticket Price: Rs. {r['ticket_price']}")
            print(f"Available Seats: {r['available_seats']}")
            print("-" * 70)

            found = True

    if found == False:
        print("No routes found.")


# Test search
# search_route(routes, "Hyderabad")
```

# BOOK A TICKET
### FEATURE 05

```python
def book_ticket(route_list, booking_list, issued_ids,
                passenger_name, passenger_age_input,
                route_id_input, seat_qty_input):

    if passenger_name == "":
        print("Passenger name is required.")
        return

    if not passenger_age_input.isdigit() or int(passenger_age_input) <= 0:
        print("Enter a valid passenger age.")
        return

    if not seat_qty_input.isdigit() or int(seat_qty_input) <= 0:
        print("Enter a valid number of seats.")
        return

    found = False

    for r in route_list:
        if r["route_id"] == route_id_input.upper():
            found = True

            seats = int(seat_qty_input)

            if seats <= r["available_seats"]:
                total_fare = r["ticket_price"] * seats
                r["available_seats"] -= seats

                booking_id = generate_booking_id(issued_ids)

                booking = {
                    "booking_id": booking_id,
                    "passenger_name": passenger_name,
                    "passenger_age": int(passenger_age_input),
                    "route_id": r["route_id"],
                    "seats": seats,
                    "total_fare": total_fare,
                    "status": "Confirmed"
                }

                booking_list.append(booking)

                print(f"\nBooking Confirmed!")
                print(f"Booking ID: {booking_id}")
                print(f"Passenger: {passenger_name}")
                print(f"Route: {r['departure_city']} to {r['destination_city']}")
                print(f"Seats: {seats}")
                print(f"Total Fare: Rs. {total_fare}")

            else:
                print("Not enough seats available.")

            break

    if found == False:
        print("Route not found.")


# Test booking
# booking1 = book_ticket(routes, bookings, issued_booking_ids,
#                        "Sarim", "21", "R101", "2")
```

# CANCEL A BOOKING
### FEATURE 07

```python
def cancel_booking(route_list, booking_list, booking_id_input):
    for b in booking_list:
        if b["booking_id"] == booking_id_input.upper():

            if b["status"] == "Cancelled":
                print("Booking is already cancelled.")
                return False

            for r in route_list:
                if r["route_id"] == b["route_id"]:
                    r["available_seats"] += b["seats"]
                    b["status"] = "Cancelled"

                    print(f"Booking ID: {b['booking_id']}")
                    print("Booking cancelled successfully.")
                    print(f"Seats restored: {b['seats']}")
                    print(f"Available seats: {r['available_seats']}")
                    return True

    print("Booking ID not found.")
    return False

```

# VIEW SEATS AVALIABLITY
### FEATURE 08

```python
def view_seat_availability(route_list, filter_available_only=False):
    print("\nSEAT AVAILABILITY")
    print("-" * 40)


    for route in route_list:
        available = route["available_seats"]
        total = route["total_seats"]

        if available == total:
            status = "Available"
        elif available == 0:
            status = "Fully Booked"
        else:
            status = "Partially Booked"


        print(f"Route: {route['route_id']} | "
              f"From: {route['departure_city']} | "
              f"To: {route['destination_city']} | "
              f"Seats: {available}/{total} | "
              f"Status: {status}")

    print("-" * 40)
```

# GENERATE BOOKING REPORT
### FEATURE 09

```python

def generate_booking_report(route_list, booking_list):
    total_created = len(booking_list)
    confirmed_count = 0
    cancelled_count = 0
    seats_reserved = 0
    confirmed_revenue = 0

    for booking in booking_list:
        if booking["status"] == "Confirmed":
            confirmed_count += 1
            seats_reserved += booking["seats"]
            confirmed_revenue += booking["total_fare"]
        elif booking["status"] == "Cancelled":
            cancelled_count += 1

    fully_booked_count = 0

    for route in route_list:
        if route["available_seats"] == 0:
            fully_booked_count += 1

    print("\nBOOKING REPORT")
    print("-" * 40)
    print(f"Total bookings: {total_created}")
    print(f"Confirmed bookings: {confirmed_count}")
    print(f"Cancelled bookings: {cancelled_count}")
    print(f"Seats reserved: {seats_reserved}")
    print(f"Confirmed revenue: Rs. {confirmed_revenue:,}")
    print(f"Fully booked routes: {fully_booked_count}")

    # print("-" * 40)

    # return {
    #     "total_created": total_created,
    #     "confirmed_count": confirmed_count,
    #     "cancelled_count": cancelled_count,
    #     "seats_reserved": seats_reserved,
    #     "confirmed_revenue": confirmed_revenue,
    #     "fully_booked_routes": fully_booked_count
    # }


generate_booking_report(routes, bookings)
```

# ROUTE SCHEDULE
### FEATURE 10

```python

def view_route_schedule(route_list, supported_cities):
    print("\nROUTE SCHEDULE BY DEPARTURE CITY")
    print("-" * 40)

    for city in supported_cities:
        print(f"\n{city}")
        has_departures = False

        for route in route_list:
            if route["departure_city"].lower() == city.lower():
                print(f"  {route['route_id']} -> "
                      f"{route['destination_city']} | "
                      f"{route['departure_time']}")
                has_departures = True

        if has_departures == False:
            print("  No scheduled departures.")

    print("-" * 40)


# view_route_schedule(routes, SUPPORTED_CITIES)
```

```python
def smart_bus():
    app_routes = get_initial_routes()
    app_bookings = []
    app_issued_ids = set()

    print("=" * 45)
    print("   Smart Bus Reservation System")
    print("=" * 45)

    while True:
        print("\n" + "=" * 30)
        print("Choose your option (1-9)")
        print("=" * 30)
        print("1. View Available Routes")
        print("2. Search for a Route")
        print("3. Book a Ticket")
        print("4. View My Bookings")
        print("5. Cancel a Booking")
        print("6. View Seat Availability")
        print("7. Generate Booking Report")
        print("8. Exit")
        print("9. View Route Schedule")
        print("=" * 30)

        choice = input("Enter your option: ").strip()

        if choice == "1":
            filter_choice = input(
                "Show routes with available seats only? (y/n): "
            ).strip().lower()

            view_available_routes(
                app_routes,
                filter_choice == "y"
            )

        elif choice == "2":
            query = input("Enter route ID or city: ")
            search_route(app_routes, query)

        elif choice == "3":
            print("\n--- Book a Ticket ---")

            p_name = input("Enter passenger name: ")
            p_age = input("Enter passenger age: ")
            r_id = input("Enter route ID (e.g. R101): ")
            s_qty = input("Enter number of seats: ")

            book_ticket(
                app_routes,
                app_bookings,
                app_issued_ids,
                p_name,
                p_age,
                r_id,
                s_qty
            )

        elif choice == "4":
            view_my_bookings(app_bookings)

        elif choice == "5":
            b_id = input("Enter booking ID to cancel: ")
            cancel_booking(app_routes, app_bookings, b_id)

        elif choice == "6":
            filter_choice = input(
                "Show routes with available seats only? (y/n): "
            ).strip().lower()

            view_seat_availability(
                app_routes,
                filter_choice == "y"
            )

        elif choice == "7":
            generate_booking_report(app_routes, app_bookings)

        elif choice == "8":
            print("Thank you for choosing Smart Bus Reservation System!")
            break

        elif choice == "9":
            view_route_schedule(app_routes, SUPPORTED_CITIES)

        else:
            print("\n*********** Invalid Choice ***********")

```
